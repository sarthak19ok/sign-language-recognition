# 07 · LSTM and GRU comparison

Compares LSTM and GRU inputs using signer-grouped cross-validation and an inner signer-held-out early-stopping subset. After choosing the hand, angle, and missingness-mask representation by cross-validation, it fits both models and evaluates the frozen test split once.

## Signer-grouped feature and model search

This comparison uses training data only for grouped cross-validation and uses validation data for a separate check.


```python
from pathlib import Path
import runpy

ROOT = Path.cwd().resolve()
while not (ROOT / "data/processed/WLASL100/all_metadata.csv").exists() and ROOT.parent != ROOT:
    ROOT = ROOT.parent
script = ROOT / "scripts" / "train_sequence_variants.py"
if not script.exists():
    raise FileNotFoundError(script)
runpy.run_path(str(script), run_name="__main__")
```

## Final LSTM and GRU results

Uses the best input variant by grouped cross-validation and writes one result CSV and checkpoint per model under `out/`. This opens the test split once for the final report of model performance.


```python
from pathlib import Path
import runpy

ROOT = Path.cwd().resolve()
while not (ROOT / "data/processed/WLASL100/all_metadata.csv").exists() and ROOT.parent != ROOT:
    ROOT = ROOT.parent
script = ROOT / "scripts" / "final_sequence_evaluation.py"
if not script.exists():
    raise FileNotFoundError(script)
runpy.run_path(str(script), run_name="__main__")
```

Each recurrent model folder in `out/` contains one `results.csv` and a saved model checkpoint. The full feature and model search is available in `data/processed/WLASL100/sequence_model_search/`.
