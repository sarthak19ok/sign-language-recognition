# 08 · Validation error analysis

Looks at per-class recall and common confusions in the validation predictions.

```python
from pathlib import Path
import runpy

ROOT = Path.cwd().resolve()
while not (ROOT / "data/processed/WLASL100/all_metadata.csv").exists() and ROOT.parent != ROOT:
    ROOT = ROOT.parent
script = ROOT / "scripts" / "analyze_validation_errors.py"
if not script.exists():
    raise FileNotFoundError(script)
runpy.run_path(str(script), run_name="__main__")
```

Outputs are saved under `data/processed/WLASL100/error_analysis/`, including per-class metrics, common confusion pairs, and plots. Interpret recall alongside support because some validation classes have few clips.

## Inspect the validation results

Start with the per-model scores and common confusions. The plots highlight which gloss pairs are most often mixed up and which classes have the lowest recall.

```python
import json
import pandas as pd
from IPython.display import Image, display

error_dir = ROOT / 'data/processed/WLASL100/error_analysis'
summary = json.loads((error_dir / 'summary.json').read_text())
score_table = pd.DataFrame(summary).T[['n', 'accuracy', 'macro_f1', 'classes']]
display(score_table.round(3))
for model_name in summary:
    display(Image(filename=str(error_dir / f'{model_name}_top_confusions.png')))
    display(Image(filename=str(error_dir / f'{model_name}_lowest_class_recall.png')))
```
