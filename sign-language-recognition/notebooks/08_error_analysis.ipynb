# 08 · Validation error analysis

Summarizes per-class recall and common confusions for validation predictions. This notebook reads validation predictions only and does not use test results.

```python
from pathlib import Path
import runpy

ROOT = Path.cwd().resolve()
while not (ROOT / "data/processed/WLASL100/all_metadata.csv").exists() and ROOT.parent != ROOT:
    ROOT = ROOT.parent
script = ROOT / "scripts" / "analyze_validation_errors.py"
if not script.exists():
    raise FileNotFoundError(script)
runpy.run_path(str(script), run_name="__main__")
```

Outputs are saved under `data/processed/WLASL100/error_analysis/`, including per-class metrics, common confusion pairs, and plots. Interpret recall alongside support because some validation classes have few clips.
